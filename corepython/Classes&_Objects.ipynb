7 Write a class Circle with object variable radius. Add three instance methods: area()
returning π×r², perimeter() returning 2×π×r, and describe() returning a formatted
string with both values. Use math.pi for accuracy.

```python
import math


class Circle:

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

    def describe(self):
        return f"Area = {self.area():.2f}, Perimeter = {self.perimeter():.2f}"


c = Circle(5)

print(c.area())
print(c.perimeter())
print(c.describe())
```

8 Extend the Student class: add a class variable total_students = 0. Add a
@classmethod called from_csv(cls, csv_string) that parses 'Name,Age,GPA' and
returns a Student object. Add another @classmethod get_total() that returns
total_students. Test both.

```python
class Student:

    total_students = 0

    def __init__(self, name, age, gpa):
        self.name = name
        self.age = int(age)
        self.gpa = float(gpa)
        Student.total_students += 1

    @classmethod
    def from_csv(cls, csv_string):
        name, age, gpa = csv_string.split(",")
        return cls(name, age, gpa)

    @classmethod
    def get_total(cls):
        return cls.total_students


s1 = Student.from_csv("John,20,8.5")
s2 = Student.from_csv("Alice,21,9.1")

print(s1.name, s1.age, s1.gpa)
print(s2.name, s2.age, s2.gpa)

print(Student.get_total())
```

9 Write a class PasswordUtils with three @staticmethod methods:
is_strong(password) — returns True if length>=8, has a digit, and has an uppercase;
generate_hint(password) — returns first 2 chars + '***'; validate_email(email) —
returns True if '@' and '.' are present. Test all three.

```python
class PasswordUtils:

    @staticmethod
    def is_strong(password):
        has_digit = any(ch.isdigit() for ch in password)
        has_upper = any(ch.isupper() for ch in password)
        return len(password) >= 8 and has_digit and has_upper

    @staticmethod
    def generate_hint(password):
        return password[:2] + "***"

    @staticmethod
    def validate_email(email):
        return "@" in email and "." in email


print(PasswordUtils.is_strong("Python123"))
print(PasswordUtils.generate_hint("Python123"))
print(PasswordUtils.validate_email("abc@gmail.com"))
```

variable library_name, object variables (books list, member_name), __init__ with
validation (name cannot be empty), instance method add_book(title), class method
create_guest(cls) that returns a Library with name='Guest', static method
is_valid_title(title) that checks the title is a non-empty string. Demonstrate each
feature with output.

```python
class Library:

    # Class variable
    library_name = "Central Library"

    def __init__(self, member_name):
        if member_name.strip() == "":
            raise ValueError("Member name cannot be empty")

        self.member_name = member_name
        self.books = []

    # Instance Method
    def add_book(self, title):
        if Library.is_valid_title(title):
            self.books.append(title)
        else:
            print("Invalid Title")

    # Class Method
    @classmethod
    def create_guest(cls):
        return cls("Guest")

    # Static Method
    @staticmethod
    def is_valid_title(title):
        return isinstance(title, str) and title.strip() != ""


lib = Library("Rahul")

print("Library:", Library.library_name)
print("Member:", lib.member_name)

lib.add_book("Python Programming")
lib.add_book("Machine Learning")

print("Books:", lib.books)

guest = Library.create_guest()

print("Guest Member:", guest.member_name)

print(Library.is_valid_title("SQL"))
print(Library.is_valid_title(""))
```

```python
#1. Write a custom iterator that prints numbers from 1 to N.
```

```python
class Numbers:

    def __init__(self, n):
        self.n = n
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.n:
            raise StopIteration

        value = self.current
        self.current += 1
        return value


obj = Numbers(5)

for i in obj:
    print(i)
```

```python
#2. Create an iterator that returns only even numbers from a given list
```

```python
class EvenIterator:

    def __init__(self, numbers):
        self.numbers = numbers
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):

        while self.index < len(self.numbers):
            value = self.numbers[self.index]
            self.index += 1

            if value % 2 == 0:
                return value

        raise StopIteration


nums = [3, 8, 5, 10, 7, 12]

for i in EvenIterator(nums):
    print(i)
```

```python
#3. Implement an iterator that iterates over a string character by character in reverse order.
```

```python
class ReverseString:

    def __init__(self, text):
        self.text = text
        self.index = len(text) - 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < 0:
            raise StopIteration

        value = self.text[self.index]
        self.index -= 1
        return value


obj = ReverseString("Python")

for ch in obj:
    print(ch)
```

```python
#4. Write an iterator that yields elements of a list with their index (don’t use enumerate).
```

```python
class MyEnumerate:

    def __init__(self, data):
        self.data = data
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.data):
            raise StopIteration

        result = (self.index, self.data[self.index])
        self.index += 1
        return result


items = ["Apple", "Banana", "Orange"]

for index, value in MyEnumerate(items):
    print(index, value)
```

```python
def digits(number):

    for digit in str(number):
        yield int(digit)


for i in digits(987654):
    print(i)
```

```python
#6.Create a generator that yields cumulative sum of numbers in a list. Example: [1,2,3] → 1, 3, 6
```

```python
def cumulative_sum(numbers):

    total = 0

    for num in numbers:
        total += num
        yield total


nums = [1, 2, 3]

for i in cumulative_sum(nums):
    print(i)
```

```python
#8. Create an iterator that yields words from a sentence one by one.
```

```python
class WordIterator:

    def __init__(self, sentence):
        self.words = sentence.split()
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.words):
            raise StopIteration

        word = self.words[self.index]
        self.index += 1
        return word


obj = WordIterator("Python is very easy")

for word in obj:
    print(word)
```

```python
#10. Implement a generator that yields running maximum from a list Example: [3,1,4,2] → 3, 3, 4,4
```

```python
def fibonacci(n):

    a = 0
    b = 1

    for i in range(n):
        yield a
        a, b = b, a + b


for num in fibonacci(10):
    print(num)
```
