```python
""" Given a list of numbers:
[5, 10, 15, 20, 25, 30]
Perform the following in a single pipeline:
• Use map() to square each number
• Use filter() to keep only numbers divisible by 5
• Use reduce() to calculate the sum of remaining number"""
```

```python
l=[5, 10, 15, 20, 25, 30]
k=reduce(lambda x,y:x+y,filter(lambda x:x%5==0, map(lambda x:x**2,l)))
print(k)
```

```python
"""Given a list:
nums = [12, 15, 7, 18, 20, 21, 25]
Use filter() and lambda to keep numbers that are divisible by 3 OR divisible by
5 but NOT divisible by both.
Explain how the logical condition works."""
```

```python
nums = [12, 15, 7, 18, 20, 21, 25]
k=list(filter(lambda x:(x%3==0 or x%5==0) and not (x%3==0 and x%5==0 ,nums)))
#k=list(filter(lambda x:(x%3==0) ^ (x%5==0),nums))
print(k)
```
