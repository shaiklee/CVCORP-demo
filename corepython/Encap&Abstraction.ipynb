```python
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive")
        else:
            self.__balance += amount
            print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print("Withdrawn:", amount)

    def get_balance(self):
        return self.__balance


account = BankAccount("12345", 1000)

account.deposit(500)
account.withdraw(300)

print(account.get_balance())

# account.__balance = 50000
# This creates a new outside attribute; it does not change private __balance.
```

```python
class SecureFile:
    def __init__(self, content, password):
        self.__content = content
        self.__password = password
        self.__logs = []

    def read(self, password):
        if password == self.__password:
            return self.__content
        else:
            self.__logs.append("Unauthorized attempt")
            return "Access denied"


file = SecureFile("Secret information", "1234")

print(file.read("1234"))
print(file.read("wrong"))

# print(file.__logs)
# Cannot directly access private logs
```

```python
#5. Create a Product class where:
'''• price cannot be negative
• discount cannot exceed 70%
• internal final price calculation should not be directly exposed
Provide only one public method get_final_price()'''
```

```python
class Product:
    def __init__(self, price, discount):
        if price < 0:
            raise ValueError("Price cannot be negative")

        if discount < 0 or discount > 70:
            raise ValueError("Discount must be between 0% and 70%")

        self.__price = price
        self.__discount = discount

    def get_final_price(self):
        return self.__price - (self.__price * self.__discount / 100)


p = Product(1000, 20)
print(p.get_final_price())
```

```python
class Character:
    def __init__(self, max_health):
        self.__max_health = max_health
        self.__health = max_health

    def damage(self, points):
        if points > 0:
            self.__health = max(0, self.__health - points)

    def heal(self, points):
        if points > 0:
            self.__health = min(self.__max_health,
                                self.__health + points)

    @property
    def health(self):
        return self.__health


c = Character(100)

c.damage(30)
print(c.health)

c.heal(20)
print(c.health)

# c.health = 200
# Error because health has no setter.
```

```python
#11. Using abc module:
''''• Create an abstract class Shape with area(), perimeter()
• Implement Circle, Rectangle, Triangle
Demonstrate:
• why base class should NOT contain calculation logic
• what happens if a subclass fails to implement one of the methods'''
```

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


class Triangle(Shape):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def perimeter(self):
        return self.a + self.b + self.c

    def area(self):
        s = self.perimeter() / 2
        return math.sqrt(s * (s-self.a) * (s-self.b) * (s-self.c))


shapes = [
    Circle(5),
    Rectangle(10, 5),
    Triangle(3, 4, 5)
]

for shape in shapes:
    print(shape.area())
    print(shape.perimeter())
```

```python
#12. Design an abstract class PaymentGateway with:
'''• authenticate()
• pay(amount)
• refund(amount)
Implement subclasses:
• UPIPayment
• CardPayment
• NetBankingPayment
Show how abstraction helps your main program call payment methods without caring about
the payment type.'''
```

```python
from abc import ABC, abstractmethod


class PaymentGateway(ABC):

    @abstractmethod
    def authenticate(self):
        pass

    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass


class UPIPayment(PaymentGateway):
    def authenticate(self):
        print("UPI authenticated")

    def pay(self, amount):
        print("Paid", amount, "using UPI")

    def refund(self, amount):
        print("Refunded", amount, "through UPI")


class CardPayment(PaymentGateway):
    def authenticate(self):
        print("Card authenticated")

    def pay(self, amount):
        print("Paid", amount, "using Card")

    def refund(self, amount):
        print("Refunded", amount, "through Card")


class NetBankingPayment(PaymentGateway):
    def authenticate(self):
        print("Net Banking authenticated")

    def pay(self, amount):
        print("Paid", amount, "using Net Banking")

    def refund(self, amount):
        print("Refunded", amount, "through Net Banking")


def make_payment(payment_method, amount):
    payment_method.authenticate()
    payment_method.pay(amount)


make_payment(UPIPayment(), 1000)
make_payment(CardPayment(), 2000)
make_payment(NetBankingPayment(), 3000)
```

```python
#14. Create an abstract class DatabaseDriver with:
''''• connect()
• execute(query)
• close()
Implement concrete drivers:
• MySQLDriver
• PostgresDriver
• SQLiteDriver
Show how abstraction helps switch databases without rewriting main code.'''
```

```python
from abc import ABC, abstractmethod


class DatabaseDriver(ABC):

    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def execute(self, query):
        pass

    @abstractmethod
    def close(self):
        pass


class MySQLDriver(DatabaseDriver):
    def connect(self):
        print("Connected to MySQL")

    def execute(self, query):
        print("MySQL executing:", query)

    def close(self):
        print("MySQL connection closed")


class PostgresDriver(DatabaseDriver):
    def connect(self):
        print("Connected to PostgreSQL")

    def execute(self, query):
        print("PostgreSQL executing:", query)

    def close(self):
        print("PostgreSQL connection closed")


class SQLiteDriver(DatabaseDriver):
    def connect(self):
        print("Connected to SQLite")

    def execute(self, query):
        print("SQLite executing:", query)

    def close(self):
        print("SQLite connection closed")


def run_database(db, query):
    db.connect()
    db.execute(query)
    db.close()


run_database(MySQLDriver(), "SELECT * FROM users")
run_database(PostgresDriver(), "SELECT * FROM users")
run_database(SQLiteDriver(), "SELECT * FROM users")
```

```python
#16. Create an abstract class RobotCommand with:
'''• execute()
• undo()
Implement:
• PickCommand
• PlaceCommand
• MoveCommand
Demonstrate how abstraction cleanly represents commands without revealing details.'''
```

```python
from abc import ABC, abstractmethod


class RobotCommand(ABC):

    @abstractmethod
    def execute(self):
        pass

    @abstractmethod
    def undo(self):
        pass


class PickCommand(RobotCommand):
    def execute(self):
        print("Robot picks the object")

    def undo(self):
        print("Robot puts the object back")


class PlaceCommand(RobotCommand):
    def execute(self):
        print("Robot places the object")

    def undo(self):
        print("Robot removes the object")


class MoveCommand(RobotCommand):
    def execute(self):
        print("Robot moves")

    def undo(self):
        print("Robot moves back")


commands = [
    PickCommand(),
    PlaceCommand(),
    MoveCommand()
]

for command in commands:
    command.execute()
    command.undo()
```

```python
#17. Create an abstract class MLModel with:
'''• train(data)
• predict(x)
• evaluate(test_set)
Implement models:
• LinearRegressionModel- some different logic
• DecisionTreeModel – some logic
Show how a generic training loop works for any model without caring about details.'''
```

```python
from abc import ABC, abstractmethod


class MLModel(ABC):

    @abstractmethod
    def train(self, data):
        pass

    @abstractmethod
    def predict(self, x):
        pass

    @abstractmethod
    def evaluate(self, test_set):
        pass


class LinearRegressionModel(MLModel):
    def train(self, data):
        print("Training Linear Regression")

    def predict(self, x):
        return "Linear Regression prediction"

    def evaluate(self, test_set):
        print("Evaluating Linear Regression")


class DecisionTreeModel(MLModel):
    def train(self, data):
        print("Training Decision Tree")

    def predict(self, x):
        return "Decision Tree prediction"

    def evaluate(self, test_set):
        print("Evaluating Decision Tree")


def training_loop(model, data, test_set):
    model.train(data)
    print(model.predict(test_set))
    model.evaluate(test_set)


training_loop(LinearRegressionModel(), [1, 2, 3], [4, 5])
training_loop(DecisionTreeModel(), [1, 2, 3], [4, 5])
```

```python
def send_email(message):
    print("Email:", message)

def send_sms(message):
    print("SMS:", message)

def send_push(message):
    print("Push:", message)


notification_type = "email"

if notification_type == "email":
    send_email("Hello")
elif notification_type == "sms":
    send_sms("Hello")
elif notification_type == "push":
    send_push("Hello")
#As more notification types are added, the if/elif becomes messy
```

```python
#with abstraction
from abc import ABC, abstractmethod


class Notifier(ABC):

    @abstractmethod
    def send(self, message):
        pass


class EmailSender(Notifier):
    def send(self, message):
        print("Email:", message)


class SMSSender(Notifier):
    def send(self, message):
        print("SMS:", message)


class PushSender(Notifier):
    def send(self, message):
        print("Push:", message)


def notify(sender, message):
    sender.send(message)


notify(EmailSender(), "Hello")
notify(SMSSender(), "Hello")
notify(PushSender(), "Hello")
#Now the main program doesn't care which notification system is being used.
```

```python
#20. Design:
'''• Abstract base class Sensor with functions read_value() and calibrate()
• Subclasses: TemperatureSensor, PressureSensor, HumiditySensor
Encapsulate:
• internal raw sensor readings
• calibration factor
Hide all raw operations and allow only a public, clean get_reading() method.'''
```

```python
#Encapsulation + abstraction
from abc import ABC, abstractmethod


class Sensor(ABC):

    def __init__(self, raw_value, calibration_factor):
        self.__raw_value = raw_value
        self.__calibration_factor = calibration_factor

    def get_reading(self):
        return self.__raw_value * self.__calibration_factor

    @abstractmethod
    def read_value(self):
        pass

    @abstractmethod
    def calibrate(self):
        pass


class TemperatureSensor(Sensor):

    def read_value(self):
        return self.get_reading()

    def calibrate(self):
        print("Temperature sensor calibrated")


class PressureSensor(Sensor):

    def read_value(self):
        return self.get_reading()

    def calibrate(self):
        print("Pressure sensor calibrated")


class HumiditySensor(Sensor):

    def read_value(self):
        return self.get_reading()

    def calibrate(self):
        print("Humidity sensor calibrated")


temperature = TemperatureSensor(25, 1.1)
pressure = PressureSensor(100, 1.05)
humidity = HumiditySensor(60, 0.95)

print(temperature.read_value())
print(pressure.read_value())
print(humidity.read_value())

temperature.calibrate()
```

```python
from abc import ABC, abstractmethod


class Account(ABC):

    # Bank-wide interest policies
    savings_rate = 0.04
    current_rate = 0.02
    fixed_deposit_rate = 0.07

    def __init__(self, account_number, balance):
        self.account_number = account_number

        if self.validate_amount(balance):
            self.__balance = balance
        else:
            raise ValueError("Invalid opening balance")

    # Controlled access to balance
    @property
    def balance(self):
        return self.__balance

    # Deposit
    def deposit(self, amount):
        if self.validate_amount(amount):
            self.__balance += amount
        else:
            raise ValueError("Deposit amount must be positive")

    # Withdraw
    def withdraw(self, amount):
        if not self.validate_amount(amount):
            raise ValueError("Withdrawal amount must be positive")

        if amount > self.__balance:
            raise ValueError("Insufficient balance")

        self.__balance -= amount

    # Static method
    @staticmethod
    def validate_amount(amount):
        return amount > 0

    # Class method
    @classmethod
    def update_interest_policy(cls, savings=None,
                               current=None,
                               fixed_deposit=None):

        if savings is not None:
            cls.savings_rate = savings

        if current is not None:
            cls.current_rate = current

        if fixed_deposit is not None:
            cls.fixed_deposit_rate = fixed_deposit

    @abstractmethod
    def calculate_interest(self):
        pass


class SavingsAccount(Account):

    def calculate_interest(self):
        return self.balance * self.savings_rate


class CurrentAccount(Account):

    def calculate_interest(self):
        return self.balance * self.current_rate


class FixedDepositAccount(Account):

    def calculate_interest(self):
        return self.balance * self.fixed_deposit_rate


# Creating different account types
savings = SavingsAccount("S001", 50000)
current = CurrentAccount("C001", 100000)
fixed = FixedDepositAccount("FD001", 200000)


# Polymorphism
accounts = [savings, current, fixed]

for account in accounts:
    print(
        type(account).__name__,
        "Balance:", account.balance,
        "Interest:", account.calculate_interest()
    )
```

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def calculate_fare(self, distance):
        pass


class Car(Vehicle):

    def calculate_fare(self, distance):
        return 50 + distance * 15


class Bike(Vehicle):

    def calculate_fare(self, distance):
        return 20 + distance * 8


class Auto(Vehicle):

    def calculate_fare(self, distance):
        return 30 + distance * 10


class Driver:

    def __init__(self, name, vehicle):
        self.name = name
        self.vehicle = vehicle

    def __str__(self):
        return f"Driver: {self.name}, Vehicle: {type(self.vehicle).__name__}"


class Ride:

    def __init__(self, driver, distance):
        self.driver = driver
        self.vehicle = driver.vehicle
        self.distance = distance

        # Private fare
        self.__fare = self.vehicle.calculate_fare(distance)

    @property
    def fare(self):
        return self.__fare

    def __str__(self):
        return (
            f"Ride | Driver: {self.driver.name} | "
            f"Vehicle: {type(self.vehicle).__name__} | "
            f"Distance: {self.distance} km | "
            f"Fare: ₹{self.fare}"
        )


# Vehicles
car = Car()
bike = Bike()
auto = Auto()

# Drivers
driver1 = Driver("Rahul", car)
driver2 = Driver("Arun", bike)
driver3 = Driver("Ravi", auto)

# Rides
ride1 = Ride(driver1, 10)
ride2 = Ride(driver2, 10)
ride3 = Ride(driver3, 10)

print(ride1)
print(ride2)
print(ride3)
```

```python
'''3. Create:
* Abstract class PaymentMethod with pay(), validate()
* Subclasses: CardPayment, WalletPayment, UPIPayment
* Encapsulate user balance
* Use @property to control reading available funds
* Overload + operator to combine two payment methods into “split payment”
* Demonstrate polymorphism through a checkout loop.'''
```

```python
from abc import ABC, abstractmethod
class PaymentMethod(ABC):

    def __init__(self, name, balance):3
        self.name = name
        self.__balance = balance

    @property
    def available_funds(self):
        return self.__balance

    def deduct(self, amount):
        if amount > self.__balance:
            raise ValueError(f"Insufficient funds in {self.name}")

        self.__balance -= amount

    @abstractmethod
    def validate(self):
        pass

    @abstractmethod
    def pay(self, amount):
        pass

    # Operator overloading
    def __add__(self, other):
        return SplitPayment(self, other)


class CardPayment(PaymentMethod):

    def validate(self):
        return self.available_funds > 0

    def pay(self, amount):
        if not self.validate():
            return False

        self.deduct(amount)
        print(f"₹{amount} paid using Card")
        return True


class WalletPayment(PaymentMethod):

    def validate(self):
        return self.available_funds > 0

    def pay(self, amount):
        if not self.validate():
            return False

        self.deduct(amount)
        print(f"₹{amount} paid using Wallet")
        return True


class UPIPayment(PaymentMethod):

    def validate(self):
        return self.available_funds > 0

    def pay(self, amount):
        if not self.validate():
            return False

        self.deduct(amount)
        print(f"₹{amount} paid using UPI")
        return True


class SplitPayment:

    def __init__(self, payment1, payment2):
        self.payment1 = payment1
        self.payment2 = payment2

    def pay(self, amount):
        half = amount / 2

        self.payment1.pay(half)
        self.payment2.pay(half)

        print("Split payment completed")


# Creating payment methods
card = CardPayment("Card", 5000)
wallet = WalletPayment("Wallet", 3000)
upi = UPIPayment("UPI", 4000)


# Polymorphism
payments = [card, wallet, upi]

for payment in payments:
    payment.pay(1000)
```

```python
'''4. Create classes:
* Person → base
* MedicalStaff(Person)
* Doctor(MedicalStaff)
* Surgeon(Doctor)
Requirements:
* Hide sensitive data (e.g., salary, patient notes)
* Abstract method perform_duty()
* Each level overrides the method with more specific behavior
* Use super() to chain constructor calls
Demonstrate consistency across hierarchy.'''
```

```python
from abc import ABC, abstractmethod


class Person(ABC):

    def __init__(self, name, age):
        self.name = name
        self.age = age

    @abstractmethod
    def perform_duty(self):
        pass


class MedicalStaff(Person):

    def __init__(self, name, age, salary):
        super().__init__(name, age)

        # Private salary
        self.__salary = salary

    def perform_duty(self):
        print(f"{self.name} provides medical support.")


class Doctor(MedicalStaff):

    def __init__(self, name, age, salary, specialization):
        super().__init__(name, age, salary)

        self.specialization = specialization

        # Private patient notes
        self.__patient_notes = "Confidential medical information"

    def perform_duty(self):
        super().perform_duty()
        print(
            f"{self.name} diagnoses patients "
            f"as a {self.specialization}."
        )


class Surgeon(Doctor):

    def __init__(
        self,
        name,
        age,
        salary,
        specialization,
        surgery_type
    ):
        super().__init__(
            name,
            age,
            salary,
            specialization
        )

        self.surgery_type = surgery_type

    def perform_duty(self):
        super().perform_duty()
        print(
            f"{self.name} performs {self.surgery_type} surgery."
        )


# Objects
staff = MedicalStaff(
    "Ramesh",
    35,
    60000
)

doctor = Doctor("Priya",40,120000,"Cardiology")

surgeon = Surgeon("Arun",45,180000,"General Surgery","Heart")


# Polymorphism
employees = [staff, doctor, surgeon]

for employee in employees:
    print("\n", type(employee).__name__)
    employee.perform_duty()
```
