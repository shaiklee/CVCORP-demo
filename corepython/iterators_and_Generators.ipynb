```python
#1.Write a custom iterator class EvenNumbers(start, count) that yields count even numbers starting from start. Implement __iter__ and __next__ correctly. Demonstrate with a for loop and with manual next() calls.
```

```python
class EvenNumbers:
    def __init__(self, start, count):
        if start % 2 != 0:
            start += 1
        self.current = start
        self.count = count
        self.generated = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.generated == self.count:
            raise StopIteration

        value = self.current
        self.current += 2
        self.generated += 1
        return value


print("Using for loop:")
for num in EvenNumbers(3, 5):
    print(num)

print("\nUsing next():")

it = EvenNumbers(10, 3)

print(next(it))
print(next(it))
print(next(it))
```

```python
#2.Create an iterator class Countdown(n) that counts down from n to 0 (inclusive). After reaching 0, raise StopIteration. Show that calling iter() on your iterator returns itself, and manually trace three next() calls.
```

```python
class Countdown:
    def __init__(self, n):
        self.current = n

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < 0:
            raise StopIteration

        value = self.current
        self.current -= 1
        return value


c = Countdown(5)

print(iter(c) is c)

print(next(c))
print(next(c))
print(next(c))

print("\nRemaining values:")

for i in c:
    print(i)
```

```python
#3. Write a function simulate_for_loop(iterable) that replicates what Python's for loop does internally — using iter() and next() and catching StopIteration. Test it on a list, a string, and your custom EvenNumbers iterator from Q1.
```

```python
class EvenNumbers:
    def __init__(self, start, count):
        if start % 2 != 0:
            start += 1
        self.current = start
        self.count = count
        self.generated = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.generated == self.count:
            raise StopIteration

        value = self.current
        self.current += 2
        self.generated += 1
        return value


def simulate_for_loop(iterable):
    iterator = iter(iterable)

    while True:
        try:
            value = next(iterator)
            print(value)
        except StopIteration:
            break


print("List")
simulate_for_loop([1, 2, 3, 4])

print("\nString")
simulate_for_loop("Python")

print("\nEven Numbers")
simulate_for_loop(EvenNumbers(2, 5))
```

```python
#4. Write a generator function prime_generator() that yields prime numbers infinitely. Use it to print the first 10 primes. Explain why this would be impossible with a regular function returning a list for a truly infinite sequence.
```

```python
def prime_generator():
    number = 2

    while True:

        prime = True

        for i in range(2, int(number ** 0.5) + 1):
            if number % i == 0:
                prime = False
                break

        if prime:
            yield number

        number += 1


gen = prime_generator()

for i in range(10):
    print(next(gen))
```

```python
#5.Write a generator function running_total(numbers) that takes a list of numbers and yields a running cumulative total at each step. Example: [1, 3, 5, 2] yields 1, 4, 9, 11.
```

```python
def running_total(numbers):
    total = 0

    for num in numbers:
        total += num
        yield total


nums = [1, 3, 5, 2]

for value in running_total(nums):
    print(value)
```

```python
#6.Compare memory usage of a list comprehension vs a generator expression for the first 500,000 squares. Use sys.getsizeof() to measure. Also time them using time.perf_counter() to compare speed.
```

```python
import sys
import time

start = time.perf_counter()

square_list = [i * i for i in range(500000)]

end = time.perf_counter()

print("List Time:", end - start)
print("List Memory:", sys.getsizeof(square_list), "bytes")


start = time.perf_counter()

square_generator = (i * i for i in range(500000))

end = time.perf_counter()

print("Generator Time:", end - start)
print("Generator Memory:", sys.getsizeof(square_generator), "bytes")
```

```python
#7.Implement the Singleton pattern for a class AppConfig that stores a settings dictionary. Verify that ConfigA = AppConfig() and ConfigB = AppConfig() are the same object with is. Show that settings added via ConfigA are accessible from ConfigB.
```

```python
class AppConfig:

    _instance = None

    def __new__(cls):

        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.settings = {}

        return cls._instance


ConfigA = AppConfig()
ConfigB = AppConfig()

ConfigA.settings["Theme"] = "Dark"
ConfigA.settings["Language"] = "English"

print(ConfigA is ConfigB)

print(ConfigB.settings)
```

```python
#8.Create a class ShoppingCart with a list of items. Implement __len__ (number of items) and __contains__ (check if item name is in cart). Add at least 5 items and test both methods.
```

```python
class ShoppingCart:

    def __init__(self):
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def __len__(self):
        return len(self.items)

    def __contains__(self, item):
        return item in self.items


cart = ShoppingCart()

cart.add_item("Laptop")
cart.add_item("Mouse")
cart.add_item("Keyboard")
cart.add_item("Monitor")
cart.add_item("Headphones")

print("Number of items:", len(cart))

print("Laptop" in cart)
print("Mobile" in cart)
```

```python
#9.Write a class Fraction(numerator, denominator) with full arithmetic support: __add__, __sub__, __mul__, __truediv__. Simplify results using math.gcd(). Test: Fraction(1,2) + Fraction(1,3) should give Fraction(5,6).
```

```python
import math

class Fraction:

    def __init__(self, numerator, denominator):
        gcd = math.gcd(numerator, denominator)
        self.numerator = numerator // gcd
        self.denominator = denominator // gcd

    def __add__(self, other):
        n = self.numerator * other.denominator + other.numerator * self.denominator
        d = self.denominator * other.denominator
        return Fraction(n, d)

    def __sub__(self, other):
        n = self.numerator * other.denominator - other.numerator * self.denominator
        d = self.denominator * other.denominator
        return Fraction(n, d)

    def __mul__(self, other):
        return Fraction(self.numerator * other.numerator,
                        self.denominator * other.denominator)

    def __truediv__(self, other):
        return Fraction(self.numerator * other.denominator,
                        self.denominator * other.numerator)

    def __str__(self):
        return f"{self.numerator}/{self.denominator}"


f1 = Fraction(1, 2)
f2 = Fraction(1, 3)

print("Addition:", f1 + f2)
print("Subtraction:", f1 - f2)
print("Multiplication:", f1 * f2)
print("Division:", f1 / f2)
```

```python
#10.Add __eq__, __lt__, __le__, __gt__, __ge__, and __hash__ to your Fraction class. Verify that fractions can be sorted with sorted(), stored in a set, and used as dictionary keys. Test that Fraction(1,2) == Fraction(2,4) returns True.
```

```python
  import math

class Fraction:

    def __init__(self, numerator, denominator):
        gcd = math.gcd(numerator, denominator)
        self.numerator = numerator // gcd
        self.denominator = denominator // gcd

    def __eq__(self, other):
        return (self.numerator == other.numerator and
                self.denominator == other.denominator)

    def __lt__(self, other):
        return self.numerator * other.denominator < other.numerator * self.denominator

    def __le__(self, other):
        return self.numerator * other.denominator <= other.numerator * self.denominator

    def __gt__(self, other):
        return self.numerator * other.denominator > other.numerator * self.denominator

    def __ge__(self, other):
        return self.numerator * other.denominator >= other.numerator * self.denominator

    def __hash__(self):
        return hash((self.numerator, self.denominator))

    def __str__(self):
        return f"{self.numerator}/{self.denominator}"


f1 = Fraction(1, 2)
f2 = Fraction(2, 4)
f3 = Fraction(3, 4)
f4 = Fraction(1, 3)

print(f1 == f2)

fractions = [f3, f1, f4]
print("Sorted:", [str(f) for f in sorted(fractions)])

fraction_set = {f1, f2, f3}
print("Set:", [str(f) for f in fraction_set])

fraction_dict = {f1: "Half", f3: "Three Fourth"}
print(fraction_dict[f2])
```

```python
#11.Write a class Matrix(rows) that wraps a 2D list. Implement __add__ (element-wise addition) and __mul__ (scalar multiplication). Also add __len__ (number of rows) and __contains__ (check if a value exists anywhere in the matrix).
```

```python
class Matrix:

    def __init__(self, rows):
        self.rows = rows

    def __add__(self, other):
        result = []

        for i in range(len(self.rows)):
            row = []
            for j in range(len(self.rows[0])):
                row.append(self.rows[i][j] + other.rows[i][j])
            result.append(row)

        return Matrix(result)

    def __mul__(self, scalar):
        result = []

        for row in self.rows:
            result.append([value * scalar for value in row])

        return Matrix(result)

    def __len__(self):
        return len(self.rows)

    def __contains__(self, value):
        for row in self.rows:
            if value in row:
                return True
        return False

    def __str__(self):
        return str(self.rows)


m1 = Matrix([[1, 2], [3, 4]])
m2 = Matrix([[5, 6], [7, 8]])

print("Addition:", m1 + m2)
print("Scalar Multiplication:", m1 * 3)
print("Rows:", len(m1))
print(3 in m1)
print(10 in m1)
```

```python
#12.COMBINED CHALLENGE: Write a class NumberStream that is BOTH a custom
#iterator AND supports dunder arithmetic. It should: (a) iterate over a sequence of numbers given at init, (b) support + with another NumberStream (concatenate sequences), (c) support len(), (d) support in operator, (e) return a generator expression when .squares() is called. Demonstrate all features.
```

```python
class NumberStream:

    def __init__(self, numbers):
        self.numbers = numbers
        self.index = 0

    def __iter__(self):
        self.index = 0
        return self

    def __next__(self):
        if self.index >= len(self.numbers):
            raise StopIteration

        value = self.numbers[self.index]
        self.index += 1
        return value

    def __add__(self, other):
        return NumberStream(self.numbers + other.numbers)

    def __len__(self):
        return len(self.numbers)

    def __contains__(self, item):
        return item in self.numbers

    def squares(self):
        return (x * x for x in self.numbers)


stream1 = NumberStream([1, 2, 3])
stream2 = NumberStream([4, 5, 6])

combined = stream1 + stream2

print("Iterating:")
for num in combined:
    print(num)

print("Length:", len(combined))

print(4 in combined)
print(10 in combined)

print("Squares:")
for value in combined.squares():
    print(value)
```
