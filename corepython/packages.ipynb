```python
my_project/
│
├── mypackage/
│   ├── __init__.py
│   ├── student.py
│   └── teacher.py
│
└── main.py
```

```python
#student.py
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display_student(self):
        print("Student Name:", self.name)
        print("Marks:", self.marks)

    def is_pass(self):
        return self.marks >= 40
```

```python
#teacher.py
class Teacher:
    def __init__(self, name, subject):
        self.name = name
        self.subject = subject

    def display_teacher(self):
        print("Teacher Name:", self.name)
        print("Subject:", self.subject)

    def teach(self):
        print(self.name, "is teaching", self.subject)
```

```python
#main.py
from mypackage.student import Student
from mypackage.teacher import Teacher


class CollegeStudent(Student):
    def __init__(self, name, marks, course):
        super().__init__(name, marks)
        self.course = course

    def display_course(self):
        print("Course:", self.course)


# Parent class object
student1 = Student("Rahul", 85)

student1.display_student()
print("Pass:", student1.is_pass())


print()


# Another parent class object
teacher1 = Teacher("Mr. Kumar", "Python")

teacher1.display_teacher()
teacher1.teach()


print()


# Child class object
student2 = CollegeStudent("Leeyaz", 90, "Data Science")

student2.display_student()
student2.display_course()
print("Pass:", student2.is_pass())
```

```python
''' Create a package with a module containing an abstract base class (ABC).
Another module in the same package should define concrete subclasses that
implement all abstract methods. Write a driver program that imports these
classes and demonstrates polymorphism.'''
```

```python
my_project/
│
├── shapes/
│   ├── __init__.py
│   ├── shape.py
│   └── concrete_shapes.py
│
└── main.py
```

```python
#shape.py
from abc import ABC, abstractmethod


class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def display(self):
        pass
```

```python
#concrete_shapes.py
from .shape import Shape


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def display(self):
        print("This is a Circle")


class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def display(self):
        print("This is a Rectangle")
```

```python
#main.py
from shapes.concrete_shapes import Circle, Rectangle


circle = Circle(5)
rectangle = Rectangle(10, 4)


shapes = [circle, rectangle]


for shape in shapes:
    shape.display()
    print("Area:", shape.area())
    print()
```

```python
'''Create three modules:
Module A: class Animal
Module B: class Walkable
Module C: class Dog that inherits from both Animal and Walkable
Demonstrate method resolution order (MRO) by calling overridden methods and
printing the MRO.'''
```

```python
my_project/
│
├── module_a.py
├── module_b.py
├── module_c.py
└── main.py
```

```python
#moda
class Animal:

    def speak(self):
        print("Animal makes a sound")

    def eat(self):
        print("Animal is eating")
```

```python
#modb
class Walkable:

    def speak(self):
        print("Walkable speaks")

    def walk(self):
        print("Walkable can walk")
```

```python
#modc
from module_a import Animal
from module_b import Walkable


class Dog(Animal, Walkable):

    def speak(self):
        print("Dog barks")

    def walk(self):
        print("Dog walks")
```

```python
#main.py
from module_c import Dog


dog = Dog()

print("Calling methods:")
dog.speak()
dog.walk()
dog.eat()


print("\nMRO:")
for cls in Dog.mro():
    print(cls)
```

```python
''' Create a class in a module that uses private attributes and @property /
@setter decorators. Import the class into another module and show how
encapsulation protects the data while still allowing controlled access.'''
```

```python
#
my_project/
│
├── account.py
└── main.py
```

```python
class BankAccount:

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, amount):
        if amount >= 0:
            self.__balance = amount
            print("Balance updated")
        else:
            print("Balance cannot be negative")

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Amount deposited")
        else:
            print("Invalid deposit amount")
```

```python
from account import BankAccount


account = BankAccount("Leeyaz", 10000)

# Controlled access to private data
print("Current balance:", account.balance)

# Update using setter
account.balance = 15000
print("New balance:", account.balance)

# Try to set an invalid balance
account.balance = -500

# Deposit money
account.deposit(2000)
print("Final balance:", account.balance)


# Direct access to private attribute
print(account.__balance)
```

```python
''' Create a module containing two classes where one uses composition and another
uses inheritance to reuse code from a base class. Import and demonstrate the
difference between the two approaches.'''
```

```python
my_project/
│
├── vehicles.py
└── main.py
```

```python
class Engine:
    def start(self):
        print("Engine started")

    def stop(self):
        print("Engine stopped")


# Composition
class Car:
    def __init__(self, brand):
        self.brand = brand
        self.engine = Engine()

    def start_car(self):
        print(self.brand, "car:")
        self.engine.start()

    def stop_car(self):
        print(self.brand, "car:")
        self.engine.stop()


# Base class
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def display_brand(self):
        print("Brand:", self.brand)

    def move(self):
        print("Vehicle is moving")


# Inheritance
class Bike(Vehicle):
    def move(self):
        print("Bike is moving")
```

```python
from vehicles import Car, Bike


# Composition
print("COMPOSITION")
car = Car("Toyota")

car.start_car()
car.stop_car()


print()


# Inheritance
print("INHERITANCE")
bike = Bike("Honda")

bike.display_brand()
bike.move()
```
