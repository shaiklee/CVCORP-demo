```python
'''18
Open/Closed
Write a NotificationService that sends messages. Start with only EmailNotifier. Then
add SMSNotifier and PushNotifier WITHOUT modifying NotificationService.
Demonstrate OCP by showing the core class never changes.'''
```

```python

from abc import ABC, abstractmethod


# Common interface
class Notifier(ABC):

    @abstractmethod
    def send(self, message):
        pass


# First notification type
class EmailNotifier(Notifier):

    def send(self, message):
        print(f"Sending Email: {message}")


# Second notification type
class SMSNotifier(Notifier):

    def send(self, message):
        print(f"Sending SMS: {message}")


# Third notification type
class PushNotifier(Notifier):

    def send(self, message):
        print(f"Sending Push Notification: {message}")


# Core service
class NotificationService:

    def __init__(self, notifier):
        self.notifier = notifier

    def notify(self, message):
        self.notifier.send(message)


# Testing all notification types

email = NotificationService(EmailNotifier())
email.notify("Welcome to Python!")

sms = NotificationService(SMSNotifier())
sms.notify("Your OTP is 1234")

push = NotificationService(PushNotifier())
push.notify("You have a new notification!")


'''OCP means extending existing functionality by adding new classes or implementations instead of repeatedly modifying existing tested classes.'''
```

```python
'''19
Liskov Substitution
Demonstrate LSP violation: create a Bird class with fly(), then create Penguin(Bird)
that raises NotImplementedError in fly(). Show how this breaks code that expects
any Bird to fly. Then fix the hierarchy so LSP is satisfied.'''
```

```python

class Bird:
    def fly(self):
        print("Bird is flying")


class Eagle(Bird):
    def fly(self):
        print("Eagle is flying")


class Sparrow(Bird):
    def fly(self):
        print("Sparrow is flying")


class Penguin(Bird):
    def fly(self):
        raise NotImplementedError("Penguins cannot fly")


# Function expecting any Bird
def make_it_fly(bird):
    bird.fly()


eagle = Eagle()
sparrow = Sparrow()
penguin = Penguin()

make_it_fly(eagle)
make_it_fly(sparrow)

make_it_fly(penguin)
'''It expects every Bird object to have working flying behavior.

But Penguin overrides fly() and raises an error.

Therefore, substituting Penguin for Bird breaks the function.
Part B: Fixing the LSP violation
The solution is to separate flying birds from birds in general.

We create:
- Bird – Common parent class.
- FlyingBird – Represents birds that can fly.
- Eagle and Sparrow – Inherit from FlyingBird.
- Penguin – Inherits directly from Bird.'''

#corrected code

class Bird:
    def eat(self):
        print("Bird is eating")


class FlyingBird(Bird):
    def fly(self):
        print("Bird is flying")


class Eagle(FlyingBird):
    pass


class Sparrow(FlyingBird):
    pass


class Penguin(Bird):
    def swim(self):
        print("Penguin is swimming")


# Function specifically for flying birds
def make_it_fly(bird):
    bird.fly()


eagle = Eagle()
sparrow = Sparrow()
penguin = Penguin()

make_it_fly(eagle)
make_it_fly(sparrow)

penguin.eat()
penguin.swim()

'''LSP states that subclasses should be replaceable for their parent classes without changing the correctness of the program. We fix violations by designing inheritance relationships around valid behavior.'''
```

```python
'''20
Interface Segregation
Design a smart home system. Create segregated interfaces: Switchable, Dimmable,
ourable. A BasicBulb implements only Switchable. A SmartBulb implements
all three. A Fan implements Switchable and has speed() separately. Show that each
device only implements what it needs. '''
```

```python

from abc import ABC, abstractmethod


# Interface 1
class Switchable(ABC):

    @abstractmethod
    def turn_on(self):
        pass

    @abstractmethod
    def turn_off(self):
        pass


# Interface 2
class Dimmable(ABC):

    @abstractmethod
    def set_brightness(self, level):
        pass


# Interface 3
class Colourable(ABC):

    @abstractmethod
    def set_colour(self, colour):
        pass


# Basic bulb implements only Switchable
class BasicBulb(Switchable):

    def turn_on(self):
        print("Basic Bulb ON")

    def turn_off(self):
        print("Basic Bulb OFF")


# Smart bulb implements all three interfaces
class SmartBulb(Switchable, Dimmable, Colourable):

    def turn_on(self):
        print("Smart Bulb ON")

    def turn_off(self):
        print("Smart Bulb OFF")

    def set_brightness(self, level):
        print(f"Brightness set to {level}%")

    def set_colour(self, colour):
        print(f"Colour changed to {colour}")


# Fan implements only Switchable
class Fan(Switchable):

    def turn_on(self):
        print("Fan ON")

    def turn_off(self):
        print("Fan OFF")


# Creating objects
basic = BasicBulb()
smart = SmartBulb()
fan = Fan()


# Basic bulb
basic.turn_on()
basic.turn_off()

print()

# Smart bulb
smart.turn_on()
smart.set_brightness(70)
smart.set_colour("Blue")
smart.turn_off()

print()

# Fan
fan.turn_on()
fan.turn_off()



'''What would violate ISP?
Imagine we created one large interface:
class SmartDevice(ABC):
    @abstractmethod
    def turn_on(self):
        pass

    @abstractmethod
    def turn_off(self):
        pass

    @abstractmethod
    def set_brightness(self, level):
        pass

    @abstractmethod
    def set_colour(self, colour):
        pass



Now BasicBulb and Fan would be forced to implement brightness and colour methods even though they don't support those features.

That is exactly what ISP tries to prevent.
ISP achieved
Interview answer: ISP states that classes should not depend on methods they do not use. We achieve this by splitting large interfaces into smaller, specific interfaces.'''
```

```python
'''M1 Inheritance + Abstraction
INHERITANCE + ABSTRACTION: Design an abstract class Vehicle(ABC) with
abstract methods start(), stop(), and fuel_type (abstract property). Create
ElectricVehicle and PetrolVehicle as concrete classes. Then create
HybridCar(ElectricVehicle, PetrolVehicle) using multiple inheritance. Use super()
correctly at each level and print the MRO.'''
```

```python

from abc import ABC, abstractmethod


class Vehicle(ABC):
    def __init__(self):
        super().__init__()
        print("Vehicle initialized")

    @abstractmethod
    def start(self):
        print("Vehicle started")

    @abstractmethod
    def stop(self):
        print("Vehicle stopped")

    @property
    @abstractmethod
    def fuel_type(self):
        pass


class ElectricVehicle(Vehicle):
    def __init__(self):
        super().__init__()
        print("ElectricVehicle initialized")

    def start(self):
        print("Electric motor starting")
        super().start()

    def stop(self):
        print("Electric motor stopping")
        super().stop()

    @property
    def fuel_type(self):
        return "Electric"


class PetrolVehicle(Vehicle):
    def __init__(self):
        super().__init__()
        print("PetrolVehicle initialized")

    def start(self):
        print("Petrol engine starting")
        super().start()

    def stop(self):
        print("Petrol engine stopping")
        super().stop()

    @property
    def fuel_type(self):
        return "Petrol"


class HybridCar(ElectricVehicle, PetrolVehicle):
    def __init__(self):
        super().__init__()
        print("HybridCar initialized")

    @property
    def fuel_type(self):
        return "Electric + Petrol"


car = HybridCar()

print("\nFuel type:", car.fuel_type)

print("\nStarting:")
car.start()

print("\nStopping:")
car.stop()

print("\nMRO:")
for cls in HybridCar.mro():
    print(cls.__name__)
```

```python
'''M2 Polymorphism + Encapsulation
POLYMORPHISM + ENCAPSULATION: Build an Employee hierarchy: Employee →
Manager, Developer, Intern. Each has private __salary and a @property salary.
Each overrides a method monthly_cost() differently (manager gets bonus, intern
gets a cap). Write a function total_payroll(employees) that works polymorphically.'''
```

```python

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self, value):
        if value < 0:
            raise ValueError("Salary cannot be negative")
        self.__salary = value

    def monthly_cost(self):
        return self.salary


class Manager(Employee):
    def monthly_cost(self):
        bonus = self.salary * 0.20
        return self.salary + bonus


class Developer(Employee):
    def monthly_cost(self):
        allowance = self.salary * 0.10
        return self.salary + allowance


class Intern(Employee):
    def monthly_cost(self):
        return min(self.salary, 20000)


def total_payroll(employees):
    total = 0

    for employee in employees:
        cost = employee.monthly_cost()
        print(f"{employee.name}: Rs.{cost:.2f}")
        total += cost

    return total


employees = [
    Manager("Rahul", 80000),
    Developer("Arun", 50000),
    Intern("Kiran", 25000)
]

print("Total Payroll:")
print("Rs.", total_payroll(employees))
```

```python
'''M3 All Four Pillars
ALL FOUR PILLARS: Design a Media Streaming system. Create an abstract class
MediaPlayer(ABC) with abstract methods play(), pause(), stop(). Encapsulate
__volume with @property and @setter (0–100 validation). Use inheritance:
AudioPlayer and VideoPlayer extend MediaPlayer. Demonstrate polymorphism by
putting both in a list and calling play() on each.'''
```

```python

from abc import ABC, abstractmethod


class MediaPlayer(ABC):
    def __init__(self, volume=50):
        self.__volume = volume

    @property
    def volume(self):
        return self.__volume

    @volume.setter
    def volume(self, value):
        if not isinstance(value, (int, float)) or not 0 <= value <= 100:
            raise ValueError("Volume must be between 0 and 100")

        self.__volume = value

    @abstractmethod
    def play(self):
        pass

    @abstractmethod
    def pause(self):
        pass

    @abstractmethod
    def stop(self):
        pass


class AudioPlayer(MediaPlayer):
    def play(self):
        print(f"Playing audio at volume {self.volume}")

    def pause(self):
        print("Audio paused")

    def stop(self):
        print("Audio stopped")


class VideoPlayer(MediaPlayer):
    def play(self):
        print(f"Playing video at volume {self.volume}")

    def pause(self):
        print("Video paused")

    def stop(self):
        print("Video stopped")


# Creating objects
audio = AudioPlayer(70)
video = VideoPlayer(40)

# Encapsulation demonstration
audio.volume = 80
print("Audio volume:", audio.volume)

# Polymorphism demonstration
players = [audio, video]

for player in players:
    player.play()
    player.pause()
    player.stop()
```

```python
'''M4 SOLID: S + O + D
SOLID — S + O + D: Build an e-commerce order processing system.
OrderProcessor should have a single responsibility. Adding a new payment method
(Bitcoin) should require zero changes to OrderProcessor (OCP). OrderProcessor
should depend on a PaymentGateway abstraction, not a concrete class (DIP). Write
complete working code.'''
```

```python

from abc import ABC, abstractmethod


# Abstraction
class PaymentGateway(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


# Concrete payment methods
class CreditCard(PaymentGateway):
    def pay(self, amount):
        print(f"Paid Rs.{amount} using Credit Card")


class UPI(PaymentGateway):
    def pay(self, amount):
        print(f"Paid Rs.{amount} using UPI")


class PayPal(PaymentGateway):
    def pay(self, amount):
        print(f"Paid Rs.{amount} using PayPal")


# New payment method
class Bitcoin(PaymentGateway):
    def pay(self, amount):
        print(f"Paid Rs.{amount} using Bitcoin")


# Order class: stores order information
class Order:
    def __init__(self, order_id, amount):
        self.order_id = order_id
        self.amount = amount


# OrderProcessor: coordinates payment
class OrderProcessor:
    def __init__(self, payment_gateway):
        self.payment_gateway = payment_gateway

    def process_order(self, order):
        print(f"Processing Order: {order.order_id}")
        self.payment_gateway.pay(order.amount)
        print("Order completed successfully")


# Create an order
order = Order(101, 5000)

# Choose a payment method
gateway = CreditCard()

processor = OrderProcessor(gateway)
processor.process_order(order)


# Change payment method without modifying OrderProcessor
print("\nUsing Bitcoin:")
bitcoin_gateway = Bitcoin()

processor2 = OrderProcessor(bitcoin_gateway)
processor2.process_order(order)
```

```python
'''M5 Inheritance + Encapsulation + LSP
INHERITANCE + ENCAPSULATION + SOLID-L: Create a base class Shape with a
protected _color and public @property color + @setter. Subclasses Circle,
Rectangle, Triangle inherit it. Demonstrate LSP: any function that works on Shape
must work with all three without modification. Also verify the setter validates that
color is a non-empty string.'''
```

```python

from abc import ABC, abstractmethod


class Shape(ABC):
    def __init__(self, color):
        self.color = color

    @property
    def color(self):
        return self._color

    @color.setter
    def color(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Color must be a non-empty string")

        self._color = value

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2


class Rectangle(Shape):
    def __init__(self, color, length, width):
        super().__init__(color)
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


class Triangle(Shape):
    def __init__(self, color, base, height):
        super().__init__(color)
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


# This function works with any Shape subclass
def display_shape(shape):
    print("Color:", shape.color)
    print("Area:", shape.area())
    print("----------------")


shapes = [
    Circle("Red", 5),
    Rectangle("Blue", 10, 4),
    Triangle("Green", 8, 6)
]

for shape in shapes:
    display_shape(shape)


# Testing color validation
circle = Circle("Yellow", 3)

circle.color = "Purple"
print("Updated color:", circle.color)

# Uncomment to test the error:
# circle.color = ""
```

```python

'''M6 Duck Typing + Abstraction + ISP
DUCK TYPING + ABSTRACTION + SOLID-I: Build a logging system with
segregated interfaces: Loggable (log(msg)), Filterable (filter(level)), and Formattable
(format(msg)). Create ConsoleLogger(Loggable), FileLogger(Loggable, Filterable),
and PrettyLogger(Loggable, Formattable). Write a function process_log(logger, msg)
that uses duck typing — it only calls log(msg) and works with any object that has
that method.'''
```

```python

from abc import ABC, abstractmethod


# Interface 1
class Loggable(ABC):
    @abstractmethod
    def log(self, msg):
        pass


# Interface 2
class Filterable(ABC):
    @abstractmethod
    def filter(self, level):
        pass


# Interface 3
class Formattable(ABC):
    @abstractmethod
    def format(self, msg):
        pass


# ConsoleLogger only needs Loggable
class ConsoleLogger(Loggable):
    def log(self, msg):
        print("Console:", msg)


# FileLogger needs Loggable and Filterable
class FileLogger(Loggable, Filterable):
    def log(self, msg):
        with open("app.log", "a") as file:
            file.write(msg + "\n")

        print("Message saved to file")

    def filter(self, level):
        print(f"Filtering file logs by level: {level}")


# PrettyLogger needs Loggable and Formattable
class PrettyLogger(Loggable, Formattable):
    def format(self, msg):
        return f"***** {msg.upper()} *****"

    def log(self, msg):
        print(self.format(msg))


# Duck typing function
def process_log(logger, msg):
    logger.log(msg)


# Demonstration
console = ConsoleLogger()
file_logger = FileLogger()
pretty = PrettyLogger()

loggers = [console, file_logger, pretty]

for logger in loggers:
    process_log(logger, "Application started")


print("\nAdditional functionality:")
file_logger.filter("ERROR")
print(pretty.format("database connected"))


# Duck typing with an unrelated class
class CustomLogger:
    def log(self, msg):
        print("Custom:", msg)


custom = CustomLogger()
process_log(custom, "Duck typing works!")
```

```python
'''M7 GRAND CHALLENGE — All Pillars + SOLID
FULL DESIGN CHALLENGE — ALL CONCEPTS: Design a Hospital Management
System with the following constraints: (1) Patient, Doctor, Nurse share common
attributes via inheritance from Person. (2) Billing is abstract — CashBilling and
InsuranceBilling are concrete. (3) Medical records are fully encapsulated with
@property access. (4) Treatment protocols use polymorphism — each specialisation
overrides treat(patient). (5) The entire system follows SRP (separate classes per
concern) and DIP (high-level classes depend on abstractions). Write at least 8
classes and demonstrate the full system running.'''
```

```python

from abc import ABC, abstractmethod


# ----------------------------------
# 1. INHERITANCE: Common Person class
# ----------------------------------

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")


class Patient(Person):
    def __init__(self, name, age, patient_id):
        super().__init__(name, age)
        self.patient_id = patient_id
        self.__medical_record = MedicalRecord()

    @property
    def medical_record(self):
        return self.__medical_record


class Doctor(Person):
    def __init__(self, name, age, specialization):
        super().__init__(name, age)
        self.specialization = specialization


class Nurse(Person):
    def __init__(self, name, age, department):
        super().__init__(name, age)
        self.department = department

    def assist_patient(self, patient):
        print(f"Nurse {self.name} is assisting {patient.name}")


# ----------------------------------
# 2. ENCAPSULATION: Medical Records
# ----------------------------------

class MedicalRecord:
    def __init__(self):
        self.__records = []

    @property
    def records(self):
        # Return a tuple so callers cannot modify the original list
        return tuple(self.__records)

    def add_record(self, record):
        if not isinstance(record, str) or not record.strip():
            raise ValueError("Medical record cannot be empty")

        self.__records.append(record)

    def display_records(self):
        print("Medical Records:")

        for record in self.records:
            print("-", record)


# ----------------------------------
# 3. ABSTRACTION: Billing
# ----------------------------------

class Billing(ABC):

    @abstractmethod
    def calculate_bill(self, amount):
        pass

    @abstractmethod
    def process_payment(self, amount):
        pass


class CashBilling(Billing):
    def calculate_bill(self, amount):
        return amount

    def process_payment(self, amount):
        print(f"Rs.{amount} paid through cash")


class InsuranceBilling(Billing):
    def calculate_bill(self, amount):
        insurance_coverage = 0.80
        return amount * (1 - insurance_coverage)

    def process_payment(self, amount):
        print(f"Rs.{amount} paid by patient after insurance")


# ----------------------------------
# 4. ABSTRACTION + POLYMORPHISM:
# Treatment Protocols
# ----------------------------------

class TreatmentProtocol(ABC):

    @abstractmethod
    def treat(self, patient):
        pass


class CardiologyTreatment(TreatmentProtocol):
    def treat(self, patient):
        print(f"Cardiology treatment provided to {patient.name}")


class OrthopedicTreatment(TreatmentProtocol):
    def treat(self, patient):
        print(f"Orthopedic treatment provided to {patient.name}")


# ----------------------------------
# 5. DIP + SRP:
# Hospital Management System
# ----------------------------------

class HospitalManagementSystem:

    def __init__(self, billing, treatment_protocol):
        # Depend on abstractions, not concrete classes
        self.billing = billing
        self.treatment_protocol = treatment_protocol

    def treat_patient(self, patient):
        self.treatment_protocol.treat(patient)

    def generate_bill(self, patient, amount):
        final_amount = self.billing.calculate_bill(amount)

        print(f"\nBill for {patient.name}: Rs.{final_amount}")
        self.billing.process_payment(final_amount)


# ----------------------------------
# 6. RUNNING THE SYSTEM
# ----------------------------------

# Create people
patient = Patient("Rahul", 35, "P101")

doctor = Doctor("Dr. Sharma", 45, "Cardiology")

nurse = Nurse("Priya", 30, "Cardiology")

print("PATIENT DETAILS")
patient.display_info()
print("Patient ID:", patient.patient_id)

print("\nDOCTOR DETAILS")
doctor.display_info()
print("Specialization:", doctor.specialization)

print("\nNURSE DETAILS")
nurse.display_info()
nurse.assist_patient(patient)


# Add medical records
print("\nMEDICAL RECORDS")

patient.medical_record.add_record("Blood pressure: 140/90")
patient.medical_record.add_record("ECG completed")

patient.medical_record.display_records()


# Select treatment protocol and billing method
treatment = CardiologyTreatment()
billing = InsuranceBilling()

hospital = HospitalManagementSystem(
    billing,
    treatment
)

# Provide treatment
print("\nTREATMENT")
hospital.treat_patient(patient)

# Generate bill
print("\nBILLING")
hospital.generate_bill(patient, 10000)


# Demonstrate polymorphism with another protocol
print("\nCHANGING TREATMENT PROTOCOL")

hospital.treatment_protocol = OrthopedicTreatment()
hospital.treat_patient(patient)


# Demonstrate changing billing method
print("\nCHANGING BILLING METHOD")

hospital.billing = CashBilling()
hospital.generate_bill(patient, 10000)
```
