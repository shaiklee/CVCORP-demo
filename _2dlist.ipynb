```python
l = [2, 1, 4, 3, 5]
for i in range(len(l) - 1):
    u = l[i+1:]
    s = min(u)
    t = u.index(s) + (i + 1)
    if l[i] > l[t]:
        l[i], l[t] = l[t], l[i]

print(l)
```
