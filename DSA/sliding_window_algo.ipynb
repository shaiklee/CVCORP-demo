```python
from operator import index


def maxsuml(arr,k):
    n=len(arr)
    ms=cs=0
    l=0
    for r in range(n):
        cs+=arr[r]
        if r-l+1==k:
            ms=max(ms,cs)
            cs-=arr[l]
            l+=1
    return ms
l=[12,24,15,25,32,9,17]
maxsuml(l,3)
```

```python
def maxsum(arr,k):
    n=len(arr)
    ms=cs=sum(arr[:k])
    for i in range(k,n):
        cs+=arr[i]-arr[i-k]
        ms=max(ms,cs)
    return ms
l=[12,24,15,25,32,9,17]
maxsum(l,3)
```

```python
#first negative in a window of size k
#bruteforce
l=[12,-1,-7,8,-15,30,16,28]
def fnegk(arr,k):
    li=[]
    for i in range(len(arr)-k+1):
        for j in range(i,i+k):
            if l[j]<0:
                li.append(l[j])
                break
        else:
            li.append(0)
    return li
fnegk(l,3)
```

```python
def fistnegative(arr,k):
    n=len(arr)
    from collections import deque
    dq=deque()
    lt=[]
    for i in range(k):
        if arr[i]<0:
            dq.append(i)
    for i in range(k,n):
        if dq:
            lt.append(arr[dq[0]])
        else:
            lt.append(0)
        while dq and dq[0]<(i-k+1):
            dq.popleft()
        if arr[i]<0:
            dq.append(i)
    if dq:
        lt.append(arr[dq[0]])
    else:
        lt.append(0)
    return lt
l=[12,-1,-7,8,-15,30,16,28]
fistnegative(l,3)
```

```python
#count of anagrm
def ang_count(s,p):
    n=len(s)
    m=len(p)
    dic={}
    for i in p:
        if i in dic:
            dic[i]+=1
        else:
            dic[i]=1
    count=l=0
    for r in range(n):
        if s[r] in dic:
            dic[s[r]]-=1
        if r-l+1==m:
            b=True
            for val in dic.values():
                if val!=0:
                    b=False
                    break
            if b:
                count+=1
            if s[l] in dic:
                dic[s[l]]+=1
            l+=1
    return count
s='forxxorfoxdofr'
ang_count(s,'for')
```

```python
#max no of vowel substring count
def maxovel(s,k):
    n=len(s)
    st={'a', 'e', 'i', 'o', 'u'}
    c=0
    for i in range(k):
        if s[i] in st:
            c+=1
    maxc=c
    for i in range(k,n):
        if s[i] in st:
            c+=1
        if s[i-k] in st:
            c-=1
        maxc=max(maxc,c)
    return maxc
s='abcaiedef'
maxovel(s,3)
# s=(1,2,3,4,5)
```

```python
#min size subarray sum
def minlen(arr,t):
    n=len(arr)
    cs=0
    minl=n+1
    l=0
    for r in range(n):
        cs+=arr[r]
        while cs>=t:
            minl=min(minl,r-l+1)
            cs-=arr[l]
            l+=1
    if minl==n+1:
        return 0
    return minl
l=[2,3,1,2,4,3]
minlen(l,7)
```

```python
#find equilibrium index or find pivot point index
def pivot(arr):
    n=len(arr)
    pre=[0]*n
    pre[0]=arr[0]
    for i in range(1,n):
        pre[i]=pre[i-1]+arr[i]
    suf=[0]*n
    suf[n-1]=arr[n-1]
    for i in range(n-2,-1,-1):
        suf[i]=suf[i+1]+arr[i]
    for i in range(1,n):
        if pre[i]==suf[i]:
            return i
#more efficient approach
def pivoteff(arr):
    n=len(arr)
    ss=sum(arr)
    ps=0
    for i in range(n):
        ss-=arr[i]
        if ps==ss:
            return i
        ps+=arr[i]
    return -1
arr=[1,7,3,6,5,6]
pivot(arr)
```
