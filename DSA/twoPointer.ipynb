```python
def revarr(arr):
    n=len(arr)
    rev=[0]*n
    for i in range(n):
        rev[i]=arr[n-i-1]
    for i in range(n):
        arr[i]=rev[i]
    return arr
l=[5,4,3,2,1]
revarr(l)
```

```python
#check for palindrome
def palindrome(arr):
    n=len(arr)
    l=0
    while l<n:
        if arr[l]==arr[n-1]:
            l+=1
            n-=1
        else:
            return False
    return True
l=[1,2,3,2,1]
palindrome(l)
```

```python
def mostwater(arr):
    l=0
    r=len(arr)-1
    mc=0
    while l<r:
        cap=(r-l)*min(arr[l],arr[r])
        mc=max(mc,cap)
        if arr[l]<=arr[r]:
            l+=1
        else:
            r-=1
    return mc
l=[1,8,6,2,5,4,8,3,7]
mostwater(l)
```

```python
#BruteForce
def mostwater(arr):
    n=len(arr)
    cap=0
    for i in range(n):
        lm=rm=arr[i]
        for j in range(0,i):
            lm=max(lm,arr[j])
        for j in range(i,n):
            rm=max(rm,arr[j])
        cap+=-arr[i]+min(lm,rm)
    return cap
l=[2,4,3,2,7,2,4,3,6,1,4,3,2,3]
mostwater(l)
```

```python
#using two pointer
def mostwater2po(arr):
    n=len(arr)
    cap=0
    lm,rm,l,r=arr[0],arr[n-1],1,n-2
    while l<=r:
        if lm<=rm:
            if arr[l]<lm:
                cap+=lm-arr[l]
            else:
                lm=arr[l]
            l+=1
        else:
            if arr[r]<rm:
                cap+=rm-arr[r]
            else:
                rm=arr[r]
            r-=1
    return cap
mostwater2po(l)
```
