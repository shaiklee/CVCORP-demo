```python
#two pointer array approach or reversing an array
def revarr(arr):
    n=len(arr)
    l=0
    r=n-1
    while l<r:
        arr[l],arr[r]=arr[r],arr[l]
        l+=1
        r-=1
    return arr
l=[5,4,3,2,1]
revarr(l)
```

```python
def revarr(arr):
    n=len(arr)
    rev=[0]*n
    for i in range(n):
        rev[i]=arr[n-i-1]
    for i in range(n):
        arr[i]=rev[i]
    return arr
l=[5,4,3,2,1]
revarr(l)
```

```python
#check for palindrome
def palindrome(arr):
    n=len(arr)
    l=0
    while l<n:
        if arr[l]==arr[n-1]:
            l+=1
            n-=1
        else:
            return False
    return True
l=[1,2,3,2,1]
palindrome(l)
```

```python
def pairsum(arr,k):
    s=set()
    for i in arr:
        com=k-i
        if com in s:
            return True
        s.add(i)
    return False
arr=[2,7,11,15]
k=9
pairsum(arr,k)
```

```python
def pairsum(arr,k):
    arr.sort()
    l=0
    r=len(arr)-1
    while l<r:
        if arr[l]+arr[r]==k:
            return True
        elif arr[l]+arr[r]<k:
            l+=1
        else:
            r-=1
    return False
arr=[2,7,11,15]
k=9
pairsum(arr,k)
```

```python
#merging two sorted arrays
def merge(a,b):
    n=len(a)
    m=len(b)
    mer=[]
    li=ri=0
    while li<n and ri<m:
        if a[li]<b[ri]:
            mer.append(a[li])
            li+=1
        else:
            mer.append(b[ri])
            ri+=1
    while li<n:
        mer.append(a[li])
        li+=1
    while ri<m:
        mer.append(b[ri])
        ri+=1
    for i in range(n):
        a[i]=mer[i]
    for i in range(m):
        b[i]=mer[n+i]
    print(a)
    print(b)
    print(mer)
a=[3,9,17,25,32]
b=[5,7,12,23,27,29]
merge(a,b)
```

```python
#Container With Most Water
def mostwater(arr):
    l=0
    r=len(arr)-1
    mc=0
    while l<r:
        cap=(r-l)*min(arr[l],arr[r])
        mc=max(mc,cap)
        if arr[l]<=arr[r]:
            l+=1
        else:
            r-=1
    return mc
l=[1,8,6,2,5,4,8,3,7]
mostwater(l)
```

```python
#using two pointer
def mostwater2po(arr):
    n=len(arr)
    cap=0
    lm,rm,l,r=arr[0],arr[n-1],1,n-2
    while l<=r:
        if lm<=rm:
            if arr[l]<lm:
                cap+=lm-arr[l]
            else:
                lm=arr[l]
            l+=1
        else:
            if arr[r]<rm:
                cap+=rm-arr[r]
            else:
                rm=arr[r]
            r-=1
    return cap
mostwater2po(l)
```
