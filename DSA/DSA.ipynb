```python
a='characters'
b='alphabets'
a=a.lower()
b=b.lower()
c=[]
d=[]
for i in a:
    if (i not in b) and (i not in c):
        c.append(i)
for i in b:
    if (i not in a) and (i not in c):
        d.append(i)
c=set(c)
d=set(d)
c.update(d)
print(c)
k=list(c)
print(*k)
```

```python
def uncommonchar(s,p):
    l=[0]*26
    for i in s:
        l[ord(i)-97]=1
    for i in p:
        if l[ord(i)-97]==1 or l[ord(i)-97]==-1:
            l[ord(i)-97]=-1
        else:
            l[ord(i)-97]=2
    s=''
    for i in range(26):
        if l[i]==1 or l[i]==2:
            s+=chr(i+97)
    return s
print(uncommonchar('characters','alphabets'))
```
