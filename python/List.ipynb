```python
#22. Write a program to perform binary search on a sorted list.
```

```python
L=list(map(int,input().split()))
n=int(input())
b=False
l,r=0,len(L)-1
while l<=r:
    m=(l+r)//2
    if L[m]==n:
        b=True
        break
    elif n>L[m]:
        l=m+1
    else:
        r=m-1
if b:
    print("Found")
else:
    print("Not Found")
```

```python
#40. Write a program to print all possible subsequences of a list.
```

```python
l=[10,20,30,40]
for i in range(1,2**(len(l))):
    ind=0
    while i>0:
        r=i%2
        if r==1:
            print(l[ind],end=' ')
        ind+=1
        i=i//2
    print()
```
