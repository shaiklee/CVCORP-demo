```python
#1. Bank Management System
'''Create a Bank class with:
• balance variable
• deposit()
• withdraw()
• check_balance()
Create a User class that inherits Bank and displays the user's name. Perform
deposit, withdrawal, and balance check.
```

```python
class Bank:
    def __init__(self):
        self.balance = 0

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print("Current Balance:", self.balance)


class User(Bank):
    def __init__(self, name):
        super().__init__()
        self.name = name

    def display(self):
        print("User Name:", self.name)


u = User("Leeyaz")
u.display()
u.deposit(5000)
u.withdraw(1500)
u.check_balance()
```

```python
#2. Employee Salary System
'''Create an Employee class with:
• emp_name
• salary
• display_details()
Create a Manager class that inherits Employee and adds a bonus(). Display the
total salary.
```

```python
class Employee:
    def __init__(self, emp_name, salary):
        self.emp_name = emp_name
        self.salary = salary

    def display_details(self):
        print("Employee Name:", self.emp_name)
        print("Salary:", self.salary)


class Manager(Employee):
    def __init__(self, emp_name, salary, bonus):
        super().__init__(emp_name, salary)
        self.bonus_amount = bonus

    def bonus(self):
        total = self.salary + self.bonus_amount
        print("Bonus:", self.bonus_amount)
        print("Total Salary:", total)


m = Manager("Rahul", 50000, 10000)
m.display_details()
m.bonus()
```

```python
#4. Food Ordering System Using Multilevel Inheritance
'''Class 1: Restaurant
• Create a method menu(item) that returns the price of the selected food
item.
Class 2: FoodCourt (inherits Restaurant)
Create the following methods:
• display_menu() – Display the available food items.
• order() – Accept the food item from the user and allow multiple orders.
• billing() – Display the total bill and add a packing charge of ₹20.
Class 3: Customer (inherits FoodCourt)
• Create an object of the Customer class.
• Call the order() method.
```

```python
class Restaurant:
    def menu(self, item):
        food = {
            "Burger": 120,
            "Pizza": 250,
            "Dosa": 80,
            "Biryani": 200
        }
        return food.get(item, 0)


class FoodCourt(Restaurant):
    def __init__(self):
        self.total = 0

    def display_menu(self):
        print("----- MENU -----")
        print("Burger  - ₹120")
        print("Pizza   - ₹250")
        print("Dosa    - ₹80")
        print("Biryani - ₹200")

    def order(self):
        self.display_menu()

        while True:
            item = input("Enter food item: ")
            price = self.menu(item)

            if price == 0:
                print("Item not available")
            else:
                self.total += price
                print(item, "added.")

            ch = input("Do you want another item? (yes/no): ")
            if ch.lower() != "yes":
                break

        self.billing()

    def billing(self):
        print("\nFood Bill:", self.total)
        print("Packing Charge: ₹20")
        print("Total Amount:", self.total + 20)


class Customer(FoodCourt):
    pass


c = Customer()
c.order()
```

```python
#5. Movie Ticket Booking System Using Multilevel Inheritance
'''Class 1: Movie
• Create a method ticket(movie) that returns the ticket price.
Class 2: Booking (inherits Movie)
Create the following methods:
• movies() – Display the available movies.
• selection() – Allow the user to book multiple tickets.
• billing() – Display the total amount and add a booking charge of ₹30.
Class 3: Customer (inherits Booking)
• Create an object and call the selection() method.
```

```python
class Movie:
    def ticket(self, movie):
        movies = {
            "Pushpa": 200,
            "Coolie": 250,
            "War2": 300,
            "Saiyaara": 180
        }
        return movies.get(movie, 0)


class Booking(Movie):
    def __init__(self):
        self.total = 0

    def movies(self):
        print("----- MOVIES -----")
        print("Pushpa  - ₹200")
        print("Coolie  - ₹250")
        print("War2    - ₹300")
        print("Saiyaara- ₹180")

    def selection(self):
        self.movies()

        while True:
            movie = input("Enter movie name: ")
            price = self.ticket(movie)

            if price == 0:
                print("Movie not available")
            else:
                tickets = int(input("Enter number of tickets: "))
                self.total += price * tickets
                print("Tickets Booked")

            ch = input("Book another movie? (yes/no): ")
            if ch.lower() != "yes":
                break

        self.billing()

    def billing(self):
        print("\nTicket Amount:", self.total)
        print("Booking Charge: ₹30")
        print("Total Amount:", self.total + 30)


class Customer(Booking):
    pass


c = Customer()
c.selection()
```

```python
#6. Online Course Enrollment System Using Multilevel Inheritance
'''Class 1: Course
• Create a method fee(course) that returns the course fee.
Class 2: Academy (inherits Course)
Create the following methods:
• courses() – Display available courses.
• enroll() – Allow the user to enroll in multiple courses.
• billing() – Display the total fee and add a registration fee of ₹100.
Class 3: Student (inherits Academy)
• Create an object and call the enroll() method.
```

```python
class Course:
    def fee(self, course):
        courses = {
            "Python": 5000,
            "Java": 6000,
            "Data Science": 8000,
            "SQL": 4000
        }
        return courses.get(course, 0)


class Academy(Course):
    def __init__(self):
        self.total = 0

    def courses(self):
        print("\nAvailable Courses")
        print("Python - ₹5000")
        print("Java - ₹6000")
        print("Data Science - ₹8000")
        print("SQL - ₹4000")

    def enroll(self):
        self.courses()

        while True:
            course = input("Enter course name: ")
            price = self.fee(course)

            if price == 0:
                print("Course not available")
            else:
                self.total += price

            ch = input("Enroll another course (yes/no): ")
            if ch.lower() != "yes":
                break

        self.billing()

    def billing(self):
        print("\nCourse Fee:", self.total)
        print("Registration Fee: ₹100")
        print("Total Amount:", self.total + 100)


class Student(Academy):
    pass


s = Student()
s.enroll()
```

```python
#7. Cab Booking System Using Hierarchical Inheritance
'''Class 1: Cab
• Create methods to calculate the fare for Bike, Auto, and Car rides.
Class 2: Uber (inherits Cab)
• Create the methods menu(), booking(), and billing().
• Add 10% GST and apply a 15% discount if the bill is above ₹1000.
Class 3: Ola (inherits Cab)
• Create the methods menu(), booking(), and billing().
• Add 12% GST and apply a 20% discount if the bill is above ₹1500.
Driver Code
• Ask the user to choose Uber or Ola and call the booking() method.
```

```python
class Cab:
    def bike(self, km):
        return km * 10

    def auto(self, km):
        return km * 15

    def car(self, km):
        return km * 20


class Uber(Cab):
    def menu(self):
        print("1.Bike\n2.Auto\n3.Car")

    def booking(self):
        self.menu()
        choice = int(input("Enter choice: "))
        km = int(input("Enter distance: "))

        if choice == 1:
            self.bill = self.bike(km)
        elif choice == 2:
            self.bill = self.auto(km)
        else:
            self.bill = self.car(km)

        self.billing()

    def billing(self):
        gst = self.bill * 0.10
        total = self.bill + gst

        if total > 1000:
            total -= total * 0.15

        print("Total Bill:", total)


class Ola(Cab):
    def menu(self):
        print("1.Bike\n2.Auto\n3.Car")

    def booking(self):
        self.menu()
        choice = int(input("Enter choice: "))
        km = int(input("Enter distance: "))

        if choice == 1:
            self.bill = self.bike(km)
        elif choice == 2:
            self.bill = self.auto(km)
        else:
            self.bill = self.car(km)

        self.billing()

    def billing(self):
        gst = self.bill * 0.12
        total = self.bill + gst

        if total > 1500:
            total -= total * 0.20

        print("Total Bill:", total)


choice = input("Choose Uber/Ola: ")

if choice.lower() == "uber":
    u = Uber()
    u.booking()
else:
    o = Ola()
    o.booking()
```

```python
#8. Grocery Shopping System Using Hierarchical Inheritance
'''Class 1: Grocery
• Create methods to return the price of Rice, Sugar, and Oil.
Class 2: Dmart (inherits Grocery)
• Create the methods items(), shopping(), and billing().
• Add 5% GST and apply a 10% discount if the bill is above ₹2000.
Class 3: Reliance Smart (inherits Grocery)
• Create the methods items(), shopping(), and billing().
• Add 5% GST and apply a 15% discount if the bill is above ₹2500.
Driver Code
• Ask the user to choose the supermarket and call the shopping() method.
```

```python
class Grocery:
    def rice(self):
        return 60

    def sugar(self):
        return 45

    def oil(self):
        return 150


class Dmart(Grocery):
    def items(self):
        print("1.Rice ₹60")
        print("2.Sugar ₹45")
        print("3.Oil ₹150")

    def shopping(self):
        self.total = 0
        self.items()

        while True:
            item = int(input("Enter item: "))
            qty = int(input("Quantity: "))

            if item == 1:
                self.total += self.rice() * qty
            elif item == 2:
                self.total += self.sugar() * qty
            elif item == 3:
                self.total += self.oil() * qty

            ch = input("More items (yes/no): ")
            if ch.lower() != "yes":
                break

        self.billing()

    def billing(self):
        total = self.total + self.total * 0.05

        if total > 2000:
            total -= total * 0.10

        print("Bill:", total)


class RelianceSmart(Grocery):
    def items(self):
        print("1.Rice ₹60")
        print("2.Sugar ₹45")
        print("3.Oil ₹150")

    def shopping(self):
        self.total = 0
        self.items()

        while True:
            item = int(input("Enter item: "))
            qty = int(input("Quantity: "))

            if item == 1:
                self.total += self.rice() * qty
            elif item == 2:
                self.total += self.sugar() * qty
            elif item == 3:
                self.total += self.oil() * qty

            ch = input("More items (yes/no): ")
            if ch.lower() != "yes":
                break

        self.billing()

    def billing(self):
        total = self.total + self.total * 0.05

        if total > 2500:
            total -= total * 0.15

        print("Bill:", total)


store = input("Choose Dmart/Reliance: ")

if store.lower() == "dmart":
    d = Dmart()
    d.shopping()
else:
    r = RelianceSmart()
    r.shopping()
```

```python
#9. Bus Ticket Booking System Using Hierarchical Inheritance
'''Class 1: Bus
• Create methods to return the fare for Sleeper, Semi-Sleeper, and AC
buses.
Class 2: RedBus (inherits Bus)
• Create the methods routes(), booking(), and billing().
• Add 10% GST and a reservation charge of ₹30.
Class 3: AbhiBus (inherits Bus)
• Create the methods routes(), booking(), and billing().
• Add 10% GST and a reservation charge of ₹20.
Driver Code
• Ask the user to choose the platform and call the booking() method.
```

```python
class Bus:
    def sleeper(self):
        return 800

    def semi(self):
        return 600

    def ac(self):
        return 1200


class RedBus(Bus):
    def routes(self):
        print("1.Sleeper")
        print("2.Semi-Sleeper")
        print("3.AC")

    def booking(self):
        self.routes()

        choice = int(input("Enter bus type: "))
        seats = int(input("Seats: "))

        if choice == 1:
            self.total = self.sleeper() * seats
        elif choice == 2:
            self.total = self.semi() * seats
        else:
            self.total = self.ac() * seats

        self.billing()

    def billing(self):
        total = self.total + self.total * 0.10 + 30
        print("Total:", total)


class AbhiBus(Bus):
    def routes(self):
        print("1.Sleeper")
        print("2.Semi-Sleeper")
        print("3.AC")

    def booking(self):
        self.routes()

        choice = int(input("Enter bus type: "))
        seats = int(input("Seats: "))

        if choice == 1:
            self.total = self.sleeper() * seats
        elif choice == 2:
            self.total = self.semi() * seats
        else:
            self.total = self.ac() * seats

        self.billing()

    def billing(self):
        total = self.total + self.total * 0.10 + 20
        print("Total:", total)


platform = input("Choose RedBus/AbhiBus: ")

if platform.lower() == "redbus":
    r = RedBus()
    r.booking()
else:
    a = AbhiBus()
    a.booking()
```

```python
#10. ATM System Using Multiple Inheritance
'''Class 1: SBI
• Create the methods deposit(amount) and check_balance().
Class 2: UnionBank
• Create the methods withdraw(amount) and mini_statement().
Class 3: ATM (inherits SBI and UnionBank)
• Create the methods menu() and transaction().
• Allow the user to perform banking operations.
Driver Code
• Create an object of the ATM class.
• Call the transaction() method.
```

```python
class SBI:
    def __init__(self):
        self.balance = 0

    def deposit(self, amount):
        self.balance += amount

    def check_balance(self):
        print("Balance:", self.balance)


class UnionBank:
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdraw Successful")
        else:
            print("Insufficient Balance")

    def mini_statement(self):
        print("Current Balance:", self.balance)


class ATM(SBI, UnionBank):

    def menu(self):
        print("1.Deposit")
        print("2.Withdraw")
        print("3.Check Balance")
        print("4.Mini Statement")
        print("5.Exit")

    def transaction(self):
        while True:
            self.menu()

            choice = int(input("Enter choice: "))

            if choice == 1:
                amt = int(input("Amount: "))
                self.deposit(amt)

            elif choice == 2:
                amt = int(input("Amount: "))
                self.withdraw(amt)

            elif choice == 3:
                self.check_balance()

            elif choice == 4:
                self.mini_statement()

            else:
                break


a = ATM()
a.transaction()
```

```python
#11. Paytm Application Using Multiple Inheritance
'''Write a Python program to implement a Paytm Application using multiple
inheritance.
Class 1: MobileRecharge
• Create the methods recharge_plans() and mobile_recharge().
Class 2: BusTicketBooking
• Create the methods display_buses() and book_ticket().
Class 3: ElectricityBills
• Create the methods bill_details() and pay_bill().
Class 4: Paytm (inherits MobileRecharge, BusTicketBooking, and
ElectricityBills)
Create the following methods:
• menu() – Display the available services.
• services() – Allow the user to choose and use any service (MobileRecharge, Bus Ticket Booking, or Electricity Bill Payment).
```

```python
class MobileRecharge:

    def recharge_plans(self):
        print("1.₹199")
        print("2.₹299")
        print("3.₹399")

    def mobile_recharge(self):
        self.recharge_plans()
        plan = int(input("Choose plan: "))

        if plan == 1:
            print("₹199 Recharge Successful")
        elif plan == 2:
            print("₹299 Recharge Successful")
        elif plan == 3:
            print("₹399 Recharge Successful")


class BusTicketBooking:

    def display_buses(self):
        print("1.Hyderabad-Chennai")
        print("2.Hyderabad-Bangalore")

    def book_ticket(self):
        self.display_buses()
        route = int(input("Choose Bus: "))
        seats = int(input("Seats: "))
        print(seats, "Ticket(s) Booked")


class ElectricityBills:

    def bill_details(self):
        amount = int(input("Enter Bill Amount: "))
        return amount

    def pay_bill(self):
        amount = self.bill_details()
        print("₹", amount, "Bill Paid Successfully")


class Paytm(MobileRecharge, BusTicketBooking, ElectricityBills):

    def menu(self):
        print("\n1.Mobile Recharge")
        print("2.Bus Ticket")
        print("3.Electricity Bill")
        print("4.Exit")

    def services(self):
        while True:
            self.menu()

            choice = int(input("Enter Choice: "))

            if choice == 1:
                self.mobile_recharge()

            elif choice == 2:
                self.book_ticket()

            elif choice == 3:
                self.pay_bill()

            else:
                break


p = Paytm()
p.services()
```
