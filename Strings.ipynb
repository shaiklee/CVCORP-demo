```python
s=input()
if len(s)==10:
    c=0
    for i in range(len(s)):
        if ((i>=0 and i<=2) or i==4 or i==9) and s(i).isupper():
            c=c+1
        elif i==3 and (s[i] in 'CHATPF'):
            c=c+1
        elif (i>=5 and i<=8) and s[i].isdigit():
            c=c+1
        else:
            break
    if c==10:
        print("valid")
    else:
        print("Invalid")
else:
    print("invalid")
```
