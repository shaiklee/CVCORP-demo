```python
import numpy as n

print("a2 -> ",a2,"\n","a4 ->",a4)
print(n.hstack([a2,a4]))
print(n.vstack([a2,a4]))

# splitting

print("a4 ->",a4)
print(n.array_split(a4,3))
print(n.array_split(a4,3,axis=1))
print(n.array_split(a4,3,axis=0))

# append

print("a4 -> ",a4)
print(n.append(a4,90))
print(n.append(a4,[1,2,3,4]))

# insert

print("a4 -> ",a4)
print(n.insert(a4,7,707))
print(n.insert(a4,4, [1,2],axis= 1))
print(n.insert(a4,1,[1,2,3,4,5],axis= 0))

# delete

print("a4 -> ",a4)
print(n.delete(a4,1))
print(n.delete(a4,1,axis= 1))
print(n.delete(a4,1,axis= 0))

# sorting
print("a4 -> ",a4)
print(n.sort(a4))

# search
print(n.where(a4%2==0))

# filter

print("a4 -> ",a4)
fetch = [True,False]
print(a4[fetch])
print(a4[a4>40])
print(a4[(a4>30) & (a4<50)])

# aggregate functions

print("a4 -> ",a4)
print(a4.sum())
print(a4.max())
print(a4.min())
print(a4.mean())
print(n.size(a4))
print(a4.cumsum())
print(a4.cumprod())

product_price = n.array([100,200,300,400,500])
quantity = n.array([1,2,3,4,5])

price_peritem = n.multiply(product_price,quantity)
print(price_peritem)
print("Total Amount Spent = ",price_peritem.sum())

product_oneachitem  = n.cumprod(([product_price,quantity]),axis=0)
print(product_oneachitem)
print(product_oneachitem[1].sum())

print(price_peritem.sum()==product_oneachitem[1].sum())

# statistical functions

print("a4 -> ",a4)
print(n.mean(a4))
print(n.median(a4))

print(a4.flatten())

import statistics as st

print(st.mode(a4.flatten()))

print(n.std(a4))
print(n.var(a4))

marks = n.array([56, 23, 78, 94, 82])
study_hours = n.array([5,2,7,9,8])

print(n.corrcoef(marks,study_hours))
```
